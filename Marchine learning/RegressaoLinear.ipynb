```python
from sklearn.linear_model import LinearRegression

# Orçamento de Marketing
x = [[100],[2500],[3250],[5000],[6500]]
# Lucro em vendas
y = [10_000, 20_500, 33_250, 59_000, 77_000]

modelo = LinearRegression()

# Treinando modelo
modelo.fit(x,y)

predicao = modelo.predict([[20_000]])

print(f"Predição de lucro: R$ {predicao[0]:,.2f}")
```

```python
from sklearn.linear_model import LinearRegression

X = [[10], [15], [20], [25], [30], [35], [40], [50], [55], [60]]
y = [10, 13, 14, 18, 19, 23, 24, 30, 32, 34]

modelo = LinearRegression()
modelo.fit(X, y)

print(f"Preço para 45 minutos: R$ {modelo.predict([[45]])[0]:.2f}")
```

# Resumo

A regressão linear usa os dados dos aluguéis anteriores para encontrar uma tendência entre tempo e preço.

X: duração do aluguel em minutos.
Y: preço do aluguel em reais.
O modelo aprende essa relação com os dados.
O gráfico mostra os pontos reais e a linha de tendência.
Com essa linha, podemos estimar o preço para novos tempos, como 45 minutos → R$ 27,04.

Em resumo: quanto maior o tempo de aluguel, maior tende a ser o preço.

```python
import matplotlib.pyplot as plt

plt.scatter(X, y, label="Dados")
plt.plot(X, modelo.predict(X), label="Regressão Linear")

plt.xlabel("Duração (minutos)")
plt.ylabel("Preço (R$)")
plt.title("Preço do aluguel de bicicleta")

plt.legend()
plt.show()
```
